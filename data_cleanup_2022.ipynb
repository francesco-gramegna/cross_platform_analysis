```python
import os
import numpy as np
import pandas as pd
import pycountry

# ── Helper functions ──────────────────────────────────────────

def convert_to_number(val):
    if isinstance(val, str):
        val = val.strip().replace(',', '')
        if val.endswith('M'):
            return round(float(val[:-1]) * 1_000_000)
        elif val.endswith('K'):
            return round(float(val[:-1]) * 1_000)
        elif val.endswith('B'):
            return round(float(val[:-1]) * 1_000_000_000)
    try:
        return round(float(val))
    except:
        return np.nan

country_name_map = {
    'Korea, Republic of': 'South Korea',
    'Russian Federation': 'Russia',
    'Türkiye': 'Turkey',
    'Viet Nam': 'Vietnam',
    'Palestine, State of': 'Palestine',
    'Venezuela, Bolivarian Republic of': 'Venezuela',
    'Iran, Islamic Republic of': 'Iran',
    'Syrian Arab Republic': 'Syria',
}

def clean_country(val):
    if not isinstance(val, str):
        return None
    val = val.strip()
    if val in ['-', '', 'nan']:
        return None
    return val

def code_to_full_country(code):
    if not isinstance(code, str):
        return None
    try:
        name = pycountry.countries.get(alpha_2=code.strip().upper()).name
        return country_name_map.get(name, name)
    except:
        return None

numeric_cols = ['followers', 'engagement_auth', 'engagement_avg',
                'views_avg', 'likes_avg', 'comments_avg', 'shares_avg']

def clean_file(path, is_2026=False):
    df = pd.read_csv(path)
    df.columns = df.columns.str.strip()

    # standardize country column name
    if 'country' in df.columns:
        df = df.rename(columns={'country': 'audience_country'})

    # clean country values
    if 'audience_country' in df.columns:
        if is_2026:
            df['audience_country'] = df['audience_country'].apply(code_to_full_country)
        else:
            df['audience_country'] = df['audience_country'].apply(clean_country)

    # convert numeric columns
    for col in numeric_cols:
        if col in df.columns:
            df[col] = df[col].apply(convert_to_number)

    return df

# ── Process all Kaggle 2022 files ─────────────────────────────

folder = "crude_dataset/kaggle_2022"
cleaned = {}

for file in os.listdir(folder):
    path = os.path.join(folder, file)
    df = clean_file(path)
    cleaned[file] = df
    print(f"✓ {file}: {df.shape}")

# ── Process 2026 file ─────────────────────────────────────────

df_2026 = clean_file("crude_dataset/2026/2026", is_2026=True)
cleaned['2026'] = df_2026
print(f"✓ 2026: {df_2026.shape}")
```

```python
for name, df in cleaned.items():
    for col in numeric_cols:
        if col in df.columns:
            # check if any value has a decimal part
            has_decimals = (df[col].dropna() % 1 != 0).sum()
            if has_decimals > 0:
                print(f"⚠️  {name} - {col}: {has_decimals} values still have decimals")

print("✓ check complete")
```

```python
output_folder = "cleaned_dataset"
os.makedirs(output_folder, exist_ok=True)

for name, df in cleaned.items():
    df.to_csv(f"{output_folder}/{name}.csv", index=False)
    print(f"✓ saved {name}.csv")
```
