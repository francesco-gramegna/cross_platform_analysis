```python
import numpy as np

def convert_to_number(val):
    if isinstance(val, str):
        val = val.strip().replace(',', '')
        if val.endswith('M'):
            return round(float(val[:-1]) * 1_000_000)
        elif val.endswith('K'):
            return round(float(val[:-1]) * 1_000)
        elif val.endswith('B'):
            return round(float(val[:-1]) * 1_000_000_000)
    try:
        return round(float(val))
    except:
        return np.nan

def clean_er(val):
    if not isinstance(val, str):
        return None
    val = val.strip()
    if val == '-':
        return None
    return float(val.replace('%', ''))

def split_name(val):
    if '@' in str(val):
        parts = str(val).rsplit('@', 1)
        return parts[0].strip(), '@' + parts[1].strip()
    return val, None

folders = ['all countries top', 'separate countries']

for folder in folders:
    path = f"cleaned_dataset/2024/{folder}"
    for file in os.listdir(path):
        if not file.endswith('.csv'):
            continue
        filepath = f"{path}/{file}"
        df = pd.read_csv(filepath)

        # split name and handle
        df[['name', 'handle']] = df['name'].apply(lambda x: pd.Series(split_name(x)))

        # convert numeric columns
        df['followers'] = df['followers'].apply(convert_to_number)
        df['potential_reach'] = df['potential_reach'].apply(convert_to_number)
        df['er'] = df['er'].apply(clean_er)

        df.to_csv(filepath, index=False)
        print(f"✓ {folder}/{file}")

print("\nDone!")
```

```python
import pandas as pd

snapshots = ['june', 'march', 'september', 'november', 'december']

dfs = {}
for s in snapshots:
    df = pd.read_csv(f"cleaned_dataset/2022/instagram_{s}.csv")
    dfs[s] = set(df['handle_instagram'].dropna())

# compare each snapshot to december
for s in snapshots:
    overlap = len(dfs[s] & dfs['december'])
    total = len(dfs[s])
    print(f"{s} vs december: {overlap}/{total} overlap ({100*overlap/total:.1f}%)")
```
